1. Create a Doc object from the file peterrabbit.txt


```python
# Import the spaCy library for natural language processing
import spacy

# Load the small English language model
nlp = spacy.load('en_core_web_sm')

# Open the file and read the entire text as a raw string
with open('peterrabbit.txt', 'r') as f:
    raw = f.read()  # raw is a plain Python string with no linguistic processing

# Process the raw string through the NLP pipeline to create a Doc object
# This performs tokenization, POS tagging, dependency parsing, and NER internally
doc = nlp(raw)

# Print the contents of the Doc object (appears similar to the raw text)
print(doc)

# Verify that the object type is spacy.tokens.doc.Doc and not a plain string
print(type(doc))
```

2. For every token in the third sentence, print the token text, the POS tag, the fine-grained TAG tag, and the description of the fine-grained tag.

```python

# Open the file and read the entire text as a raw string
with open('peterrabbit.txt', 'r') as f:
    raw = f.read()

# Process the raw string through the NLP pipeline to create a Doc object
doc = nlp(raw)

# Convert the Doc object into a list of sentences
sentences = list(doc.sents)

# Extract the third sentence (index 2)
third_sentence = sentences[2]

print("Third Sentence:", third_sentence)
print()

# Iterate over every token in the third sentence
for token in third_sentence:
    print(
        token.text,          # The original word/token text
        token.pos_,          # Coarse-grained POS tag (e.g., NOUN, VERB)
        token.tag_,          # Fine-grained TAG tag (e.g., NNS, VBD)
        spacy.explain(token.tag_)  # Human-readable description of the fine-grained tag
    )
```

3. Provide a frequency list of POS tags from the entire document

```python
# Import Counter to count the frequency of each POS tag
from collections import Counter

# Load the small English language model
nlp = spacy.load('en_core_web_sm')

# Open the file and read the entire text as a raw string
with open('peterrabbit.txt', 'r') as f:
    raw = f.read()

# Process the raw string through the NLP pipeline to create a Doc object
doc = nlp(raw)

# Extract POS tag for every token in the document
pos_tags = [token.pos_ for token in doc]

# Count the frequency of each POS tag using Counter
pos_freq = Counter(pos_tags)

# Print the frequency list sorted by most common first
print(f"{'POS Tag':<15} {'Description':<30} {'Frequency'}")
print("-" * 55)

for pos, freq in pos_freq.most_common():
    description = spacy.explain(pos)   # Human-readable description of the POS tag
    print(f"{pos:<15} {description:<30} {freq}")
```

4. CHALLENGE: What percentage of tokens are nouns?

```python
# Open the file and read the entire text as a raw string
with open('peterrabbit.txt', 'r') as f:
    raw = f.read()

# Process the raw string through the NLP pipeline to create a Doc object
doc = nlp(raw)

# Count the total number of tokens in the document
total_tokens = len(doc)

# Count the number of tokens where POS tag is NOUN
noun_count = len([token for token in doc if token.pos_ == 'NOUN'])

# Calculate the percentage of nouns
noun_percentage = (noun_count / total_tokens) * 100

# Print the results
print(f"Total Tokens  : {total_tokens}")
print(f"Noun Count    : {noun_count}")
print(f"Noun %        : {noun_percentage:.2f}%")
```

6. Show the first two named entities from Beatrix Potter's The Tale of Peter Rabbit

```python
 # Open the file and read the entire text as a raw string
with open('peterrabbit.txt', 'r') as f:
    raw = f.read()

# Process the raw string through the NLP pipeline to create a Doc object
doc = nlp(raw)

# Extract all named entities from the document
entities = list(doc.ents)

# Slice the first two named entities
first_two = entities[:2]

# Print the first two named entities with their label and description
print(f"{'Entity Text':<25} {'Label':<15} {'Description'}")
print("-" * 60)

for ent in first_two:
    print(f"{ent.text:<25} {ent.label_:<15} {spacy.explain(ent.label_)}")
```

8. CHALLENGE: How many sentences contain named entities?

```python
# Open the file and read the entire text as a raw string
with open('peterrabbit.txt', 'r') as f:
    raw = f.read()

# Process the raw string through the NLP pipeline to create a Doc object
doc = nlp(raw)

# Convert the Doc sentences generator into a list
sentences = list(doc.sents)

# Count sentences that contain at least one named entity
sentences_with_ents = [sent for sent in sentences if sent.ents]

# Print the results
print(f"Total Sentences          : {len(sentences)}")
print(f"Sentences with Entities  : {len(sentences_with_ents)}")
print()

# Print each sentence that contains a named entity along with its entities
print("Sentences containing Named Entities:")
print("-" * 60)

for i, sent in enumerate(sentences_with_ents, start=1):
    print(f"\nSentence {i}: {sent.text.strip()}")
    print("  Entities found:")
    for ent in sent.ents:
        print(f"    → {ent.text:<20} {ent.label_:<10} {spacy.explain(ent.label_)}")
```
